```python
expression = input('please input a expresion :')
stack = list()
left = '{[('
right = '}])'
not_match = False
show = list()
print(expression)


for i in expression:
    
    show.append(i)
    print(show)
    if i in left:
        stack.append(i)
    elif i in right:
        if stack == []:
            not_match = True
            break
        else:
            left_i = stack.pop()
            if left.index(left_i) != right.index(i):
                not_match = True
                break
if stack != [] :
    not_match = True
    
print(not_match)
```

```python
import matplotlib.pyplot as plt
import random

count = [0]*6
probability =[ [],[],[],[],[],[] ]

for _ in range(1000):
    for i in range(0,50):
        count[random.randint(0,5)] +=1
    for i in range(0,6):
        probability[i].append(count[i]/sum(count))
    
print(probability)
x= list(range(1000))
plt.plot(x,probability[0],x,probability[1], x,probability[2], x,probability[3], x,probability[4], x,probability[5])
plt.xlabel('YYY')
plt.ylabel('XXX')
#plt.show()
plt.savefig('probability.png')
plt.bar(x,probability[0])
plt.show()

    
```
