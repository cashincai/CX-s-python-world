```python
counts = dict()
for i in words:
    counts.setdefault(i,0)
    counts[i] += 1
print(counts)

```

```python
words_by_count = dict()
for i in counts.keys():
    words_by_count.setdefault(counts[i],[])
    words_by_count[counts[i]].append(i)
print(words_by_count)

```
